> 📖 **Description:**
> **Greedy Best-First Search (GBFS)** is an **informed search algorithm** that selects the node that appears to be closest to the goal based on a **heuristic function h(n)**. <br><br>

---

```python
graph = {
    'A': [('B', 12), ('C', 8), ('D', 15)],
    'B': [('E', 6), ('F', 10)],
    'C': [('F', 3), ('G', 5)],
    'D': [('G', 7)],
    'E': [('H', 2)],
    'F': [('H', 0)],
    'G': [('H', 1)],
    'H': []
}

goal = 'H'
node = 'A'
path = ['A']

while node != goal:
    print(f"Visiting: {node}")
    node = min(graph[node], key=lambda x: x[1])[0]  # choose lowest h(n)
    path.append(node)

print("Path found:", path)
```
