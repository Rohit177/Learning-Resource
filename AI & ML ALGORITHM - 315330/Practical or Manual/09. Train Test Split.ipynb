>  📖 **Description:**
In **Machine Learning**, a dataset is typically divided into **Training** and **Testing** sets before building a predictive model. <br><br>
📊 The **Training Set** is used to train the machine learning model, allowing it to learn patterns and relationships from the data. <br><br>
🧪 The **Testing Set** is used to evaluate the performance of the trained model on unseen data and measure its accuracy and generalization capability.<br><br>
>🔍 A common practice is to split the dataset in ratios such as:
> - 80% Training and 20% Testing
> - 70% Training and 30% Testing
> - 75% Training and 25% Testing

### Import Library

```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
```

### Import Dataset

```python
url = 'https://raw.githubusercontent.com/kirenz/datasets/master/cars.csv'
df = pd.read_csv(url)
df.head()
```

### Plot data

```python
plt.scatter(df['Miles_per_Gallon'],df['Horsepower'])
```

### Define X and Y

```python
X = df[['Miles_per_Gallon','Cylinders']]
Y = df['Horsepower']
```

> Shape of X and y

```python
print("Shape of X:", X.shape)
print("Shape of Y:", Y.shape)
```

### Split into train test
> test size 20%

```python
x_train, x_test,y_train,y_test = train_test_split(X,Y,test_size =0.2)
```

> Shape of test train

```python
print('x_train:', x_train.shape, 'y_train:', y_train.shape)
print('x_test: ', x_test.shape,  'y_test: ', y_test.shape)
```
